```python
import pandas as pd

train = pd.read_csv("House price datasets/train.csv")
test = pd.read_csv("House price datasets/test.csv")

print(train.shape)
print(test.shape)
```

```python
train.isnull().sum().sort_values(ascending=False).head(20)
```

```python
num_cols = train.select_dtypes(include=['int64','float64']).columns
cat_cols = train.select_dtypes(include=['object']).columns

print("Numerical:", len(num_cols))
print("Categorical:", len(cat_cols))
```

```python
train[cat_cols] = train[cat_cols].fillna("None")
test[cat_cols] = test[cat_cols].fillna("None")

train[cat_cols].isnull().sum().sum()
```

```python
# Remove target column from numerical list
num_cols = num_cols.drop("SalePrice")

train[num_cols] = train[num_cols].fillna(train[num_cols].median())
test[num_cols] = test[num_cols].fillna(test[num_cols].median())

train[num_cols].isnull().sum().sum()
```

```python
train = pd.get_dummies(train, drop_first=True)
test = pd.get_dummies(test, drop_first=True)

train.shape, test.shape
```

```python
train, test = train.align(test, join='left', axis=1, fill_value=0)

train.shape, test.shape
```

```python
X = train.drop("SalePrice", axis=1)
y = train["SalePrice"]

print(X.shape, y.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

# remove target column from test
test_features = test.drop("SalePrice", axis=1)

X_test_scaled = scaler.transform(test_features)

X_scaled.shape, X_test_scaled.shape
```

```python
import numpy as np

np.save("X_train.npy", X_scaled)
np.save("y_train.npy", y)
np.save("X_test.npy", X_test_scaled)
```
